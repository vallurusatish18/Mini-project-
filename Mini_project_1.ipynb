```python
# Import Libraries
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
# LOAD DATASET
# ============================================

# Read CSV File
df = pd.read_csv("COVID-19.csv")

# Display First 5 Rows
print(df.head())
```

```python
# Numeric Columns
numeric_columns = [
    'Total Cases',
    'New Cases',
    'Total Deaths',
    'New Deaths',
    'Total Recovered',
    'New Recovered',
    'Active Cases',
    'Serious, Critical',
    'Tot Cases/1M pop',
    'Deaths/1M pop',
    'Total Tests',
    'Tests/1M pop',
    'Population'
]
```

```python
# Remove commas and convert to numeric
for col in numeric_columns:

    df[col] = (
        df[col]
        .astype(str)
        .str.replace(',', '', regex=False)
        .replace('nan', np.nan)
    )

    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
# Fill Missing Values
df[numeric_columns] = df[numeric_columns].fillna(0)

# Dataset Info
print(df.info())
```

```python
# TOP 10 COUNTRIES BY TOTAL CASES
top_cases = df.sort_values(
    by='Total Cases',
    ascending=False
).head(10)

fig_cases = px.bar(
    top_cases,
    x='Country',
    y='Total Cases',
    color='Total Cases',
    title='Top 10 Countries by Total COVID Cases',
    text='Total Cases'
)

fig_cases.update_layout(
    xaxis_title='Country',
    yaxis_title='Total Cases',
    template='plotly_dark'
)

fig_cases.show()
```

```python
# TOP 10 COUNTRIES BY DEATHS

top_deaths = df.sort_values(
    by='Total Deaths',
    ascending=False
).head(10)

fig_deaths = px.pie(
    top_deaths,
    names='Country',
    values='Total Deaths',
    title='COVID-19 Death Distribution'
)

fig_deaths.show()
```

```python
# SCATTER PLOT
# TOTAL CASES VS TOTAL DEATHS

fig_scatter = px.scatter(
    df,
    x='Total Cases',
    y='Total Deaths',
    size='Population',
    color='Country',
    hover_name='Country',
    title='Total Cases vs Total Deaths'
)

fig_scatter.show()
```

```python
# CHOROPLETH WORLD MAP

fig_map = px.choropleth(
    df,
    locations='Country',
    locationmode='country names',
    color='Total Cases',
    hover_name='Country',
    color_continuous_scale='Reds',
    title='Global COVID-19 Cases Map'
)

fig_map.show()
```

```python
# TESTING ANALYSIS

top_testing = df.sort_values(
    by='Total Tests',
    ascending=False
).head(10)

fig_test = px.bar(
    top_testing,
    x='Country',
    y='Total Tests',
    color='Total Tests',
    title='Top Countries by Total Tests'
)

fig_test.show()
```

```python
# ACTIVE CASES ANALYSIS
top_active = df.sort_values(
    by='Active Cases',
    ascending=False
).head(10)

fig_active = px.bar(
    top_active,
    x='Country',
    y='Active Cases',
    color='Active Cases',
    title='Top Countries by Active Cases'
)

fig_active.show()
```

```python

# RECOVERY ANALYSIS
fig_recovery = px.scatter(
    df,
    x='Total Cases',
    y='Total Recovered',
    size='Population',
    hover_name='Country',
    title='Cases vs Recoveries'
)

fig_recovery.show()
```

```python
# HISTOGRAM OF TOTAL CASES

fig_hist = px.histogram(
    df,
    x='Total Cases',
    nbins=30,
    title='Distribution of Total Cases'
)

fig_hist.show()
```

```python
# COMBINED DASHBOARD

fig_dashboard = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        'Top Cases',
        'Top Deaths',
        'Top Tests',
        'Active Cases'
    )
)
```

```python
# Chart 1 - Total Cases

fig_dashboard.add_trace(
    go.Bar(
        x=top_cases['Country'],
        y=top_cases['Total Cases'],
        name='Cases'
    ),
    row=1,
    col=1
)
```

```python
# Chart 2 - Deaths

fig_dashboard.add_trace(
    go.Bar(
        x=top_deaths['Country'],
        y=top_deaths['Total Deaths'],
        name='Deaths'
    ),
    row=1,
    col=2
)
```

```python
# Chart 3 - Tests

fig_dashboard.add_trace(
    go.Bar(
        x=top_testing['Country'],
        y=top_testing['Total Tests'],
        name='Tests'
    ),
    row=2,
    col=1
)
```

```python
# Chart 4 - Active Cases
fig_dashboard.add_trace(
    go.Bar(
        x=top_active['Country'],
        y=top_active['Active Cases'],
        name='Active Cases'
    ),
    row=2,
    col=2
)
```

```python
# DASHBOARD LAYOUT

fig_dashboard.update_layout(
    title='Global COVID-19 Interactive Dashboard',
    height=800,
    width=1200,
    template='plotly_dark'
)

# Show Dashboard
fig_dashboard.show()
```

```python
# SAVE DASHBOARD

fig_dashboard.write_html("covid_dashboard.html")

print("Dashboard Created Successfully!")
```
